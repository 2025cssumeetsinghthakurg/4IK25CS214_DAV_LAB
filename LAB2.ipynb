```python
 import pandas as pd
import numpy as np
```

```python
data={
    "Name": ["Amit","Bebo","Chetan","Amit","Divya","Ravi"],
    "Age": [25, np.nan,30,25,28,22],
    "Marks":[85,90,np.nan,85,78,95],
    "City":["Mysuru","Mysuru","Bangluru","Mysuru",np.nan,"Delhi"]
}
df=pd.DataFrame(data)

print("Dataset:")
print(df)
```

```python
print("\nShape of data(rows,columns):",df.shape)
print("\nColumn names:",df.columns.tolist())
print("\nData types of each column:")
print(df.dtypes)
print("\nDataset info:")
df.info()
```

```python
print("\nDescriptive statistics (numerical columns):")
print(df.describe())

print("\nMean:\n",df.mean(numeric_only=True))
print("\nMedian:",df.median(numeric_only=True))
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in datset",df.isnull().sum().sum())
```
